```python
from pathlib import Path
import os
import pandas as pd

# Optional override for users who store the dataset somewhere else.
data_dir_override = os.getenv("HEART_DISEASE_DATA_DIR")

# Get project root
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

# Relative dataset path
data_path = (
    project_root
    / "FLamby-main"
    / "flamby"
    / "datasets"
    / "fed_heart_disease"
    / "dataset_creation_scripts"
    / "heart_disease_dataset"
)

files = {
    "cleveland": "processed.cleveland.data",
    "hungarian": "processed.hungarian.data",
    "switzerland": "processed.switzerland.data",
    "va": "processed.va.data"
}

# Check that all four files are present before loading anything.
missing_files = [
    filename for filename in files.values()
    if not (data_path / filename).is_file()
]

if missing_files:
    raise FileNotFoundError(
        f"Dataset files were not found in: {data_path.resolve()}\n"
        f"Missing files: {missing_files}\n"
        "Follow the dataset download instructions in README.md, then rerun "
        "the notebook from the project root."
    )

datasets = {}

for name, filename in files.items():
    file_path = data_path / filename
    df = pd.read_csv(file_path, header=None, na_values="?")
    datasets[name] = df
    print(f"{name}: {df.shape}")
```

### Dictionary for feature names

```python
column_names = {
    0:"age",
    1:"sex",
    2:"cp",
    3:"trestbps",
    4:"chol",
    5:"fbs",
    6:"restecg",
    7:"thalach",
    8:"exang",
    9:"oldpeak",
    10:"slope",
    11:"ca",
    12:"thal",
    13:"num"
}
```

```python
for name, df in datasets.items():
 df.columns = column_names.values()
```
