```python
from pathlib import Path
import os
import pandas as pd

# Project root: supports launching Jupyter from the repository root
# or from a "notebooks" subfolder.
project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

files = {
    "cleveland": "processed.cleveland.data",
    "hungarian": "processed.hungarian.data",
    "switzerland": "processed.switzerland.data",
    "va": "processed.va.data"
}

def contains_all_dataset_files(folder):
    """Return True only when all four expected hospital files are present."""
    return all((folder / filename).is_file() for filename in files.values())

# 1. An explicitly configured directory takes priority.
data_dir_override = os.getenv("HEART_DISEASE_DATA_DIR")

if data_dir_override:
    data_path = Path(data_dir_override).expanduser()
    # Interpret a relative override from the repository root.
    if not data_path.is_absolute():
        data_path = project_root / data_path
    data_path = data_path.resolve()

    if not contains_all_dataset_files(data_path):
        raise FileNotFoundError(
            f"HEART_DISEASE_DATA_DIR is set to {data_path}, but it does not "
            "contain all four expected .data files. Check the variable and "
            "make sure it points to the folder containing the files."
        )
else:
    # 2. Preferred location documented in README.md.
    default_data_path = (
        project_root / "data" / "heart_disease_dataset"
    ).resolve()

    # 3. Backward-compatible fallback for users who already downloaded
    # the data inside the FLamby-main source checkout.
    legacy_flamby_path = (
        project_root
        / "FLamby-main"
        / "flamby"
        / "datasets"
        / "fed_heart_disease"
        / "dataset_creation_scripts"
        / "heart_disease_dataset"
    ).resolve()

    if contains_all_dataset_files(default_data_path):
        data_path = default_data_path
    elif contains_all_dataset_files(legacy_flamby_path):
        data_path = legacy_flamby_path
        print(
            "Using the existing dataset inside FLamby-main. "
            "For the standard project layout, move/copy the four files to "
            "data/heart_disease_dataset/ or set HEART_DISEASE_DATA_DIR."
        )
    else:
        data_path = default_data_path
        raise FileNotFoundError(
            f"Dataset files were not found in the default location: {default_data_path}\n"
            f"Also checked the existing FLamby location: {legacy_flamby_path}\n"
            "Place all four .data files in data/heart_disease_dataset/ under "
            "the project root, or set HEART_DISEASE_DATA_DIR to the folder "
            "containing them. See README.md for instructions."
        )


datasets = {}

for name, filename in files.items():
    file_path = data_path / filename
    df = pd.read_csv(file_path, header=None, na_values="?")
    print(f"{name}: {df.shape}")
```

```python
numeric_cols = ["age","trestbps","chol","thalach","thal"]

def zero_check(df, name):
    result = pd.DataFrame({
        "zeros":(df[numeric_cols] == 0).sum(),
        "non_null_rows":df[numeric_cols].notna().sum(),

    })
    result["zeros_%_of_non_null"] = (result["zeros"] / result["non_null_rows"] * 100).round(1)
    print(f"{name}: zero counts in numeric columns")
    return result
```

```python
zero_check(hospital_cleveland, "Cleveland")
```


### Hidden Missing Values Analysis Cleveland Dataset

To identify hidden missing values, I checked whether numerical zeros were being used as placeholders for missing data.

I examined the zero counts and the percentage of zeros among non-null observations for the following columns:

- `age`
- `trestbps` (Resting Blood Pressure)
- `chol` (Cholesterol)
- `thalach` (Maximum Heart Rate)
- `thal` (Thalassemia)

**Observation:**

No suspicious zero values were identified in these columns. The zero-value analysis did not indicate any additional hidden missing values.

```python
def binarize_target(df, name):
    # Safety check: a NaN in num would silently become 0 with a > comparison
    assert df["num"].notna().all(), f"{name}: num contains missing values"

    print(f"{name}: num before")
    print(df["num"].value_counts().sort_index())

    df["target"] = (df["num"] > 0).astype(int)

    print(f"\n{name}: target after")
    print(df["target"].value_counts().sort_index())
    print("-" * 40)
```

```python
binarize_target(hospital_cleveland, "Cleveland")
```

```python
def hospital_summary(df, name):
    cols = [c for c in df.columns if c != "target"]

    # Per-column table: missing % and zero count for ALL columns
    table = pd.DataFrame({
        "missing_%": (df[cols].isna().mean() * 100).round(1),
        "zero_count": (df[cols] == 0).sum(),
        "zero_%": ((df[cols] == 0).sum() / len(df) * 100).round(1),
    })

    # Target class balance
    counts = df["target"].value_counts().reindex([0, 1], fill_value=0)
    balance = pd.DataFrame({
        "count": counts,
        "percent": (counts / len(df) * 100).round(1),
    })
    balance.index.name = "target"
    ratio = counts.max() / counts.min() if counts.min() > 0 else float("inf")

    print(f"===== {name} =====")
    print(f"Number of rows: {len(df)}")
    display(table)
    display(balance)
    print(f"Class imbalance (majority : minority) = {ratio:.2f} : 1\n")
```

```python
hospital_summary(hospital_cleveland, "Cleveland")
```

### Observations

- The `fbs` column has the highest zero count, with 258 zeros (85.1%).
- The `exang` column contains 204 zeros (67.3%).
- The `num` column contains 164 zeros (54.1%).
- The target variable has 164 instances of class 0 (54.1%) and 139 instances of class 1 (45.9%).
- The class imbalance ratio is 1.18:1, indicating a relatively balanced target distribution.
